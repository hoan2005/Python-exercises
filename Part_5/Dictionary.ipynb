# Times ten

```python
def times_ten(start_index: int, end_index: int):
    dic={}
    for i in range(start_index,end_index+1):
        dic[i]=i*10
    return dic
d = times_ten(3, 6)
print(d)
```

# Factorials

```python
def giai_thua(n: int):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result
def factorials(n: int):
    dic={}
    for i in range(1,n+1):
        dic[i]=giai_thua(i)
    return dic
k = factorials(5)
print(k[1])
print(k[3])
print(k[5])
```

# Histogram

```python
def histogram(strings : str):
    dic={}
    for i in strings:
        if i not in dic:
            dic[i]="*"
        else:
            dic[i]+="*"
    for i in dic:
        print(f"{i} {dic[i]}")
        
    
histogram("statistically")
```
